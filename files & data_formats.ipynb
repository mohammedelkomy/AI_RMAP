## Open () 

هي الكويري اللي بنستخدمه علشان نقرأ او نكتب في الملفات بتاعتنا 
# Syntax : 
علشان تفتح ملف محتاج مننا حاجتين : 
- 1 : اسم الملف 
- 2 : وضع الفتح (Mode) :
*         'r' >> read
*         'w' >> write >> بيمسح كل القديم خلي بالك ويكتب من الاول علي نضافه 
*         'a' >> append >> بيكتب في اخر الملف من غير ما يمسح القديم 

## write


```python

file = open(r'D:\data.txt', 'w')

file.write("Hello Elkomy, weloce")

file.close() # important 
```

 EXAMPLE (WRITE) - GEMENI

```python
file = open(r'D:\data.txt', 'w')

file.write('...step 1: Extracting data \n Step 2: Data extracted successfully.')

file.close()
```

- بس دي طريقه لو حصل ايرور في الكود كده الملف مش هيتقفل ودي طبعا هي الكريقه القديمه 
- الطريقه الجديده هي استخدام with

```python
with open(r'D:\data.txt', 'w') as file :
    file.write('...step 1: Extracting data \n Step 2: Data extracted successfully.')

print('File Closed Automatically')
```

## read


```python
# Good with small files >> output is string

file = open(r'D:\data.txt', 'r')

contenet = file.read()
print(contenet)

file.close()
```

```python
# readlines best for filtering rows >> output is list

with open(r'D:\data.txt', 'r') as file:
    lines = file.readlines()

print(lines) 

# filter

print(lines[0].strip())
```

```python
# dataengineers >> best for big data or big files >> use for loop + with 

with open (r'D:\data.txt', 'r', encoding='utf-8') as file:
    for line in file:
        print(line)
```

# append

```python
with open(r'D:\data.txt', 'a') as file :
    file.write('all done...')

print('File Closed Automatically')
```

# CSV File
------------------------------


first : import csv
- reading CSVs :
-     'csv.reader' >> retuen each row as list
-     'csv.dictreader' >> retuen each row as dictionary >> best for DE

```python
import csv

with open('employees.csv', 'r', encoding='utf-8') as file:
    csv_data = csv.DictReader(file)
    
    for row in csv_data:
        print(f"Employee: {row['Name']} works in {row['Department']}")
```

```python
import csv

with open(r"D:\Excel Projects\Data\Fast Food Sales.csv", 'r', encoding='utf-8') as file:
    data = csv.DictReader(file)
    for row in data:
        # print(row)
        # دلوقتي تقدر توصل لأي معلومة باسم العمود!
        # print(row['order'])
        print(f"id is: {row['order_id']} and date is: {row['date']}")
```

first : import csv
- writing CSVs :
-       'csv.dictwriter' 
- هنا لازم تعرف بايثون الاول اسماء العواميد اللي هتكتب فيها 

```python
new_data = {
    'order_id' : 1001,
    'date' : 7/27/2026,
    'item_name' : 'laptop'
}

headers = ['order_id', 'date', 'item_name']
```

```python
import csv 

with open(r"D:\Excel Projects\Data\Fast Food Sales.csv", 'w', encoding='utf-8') as file:
    writer = csv.DictWriter(file, fieldnames=headers)
    writer.writeheader()
    writer.writerow(new_data)

# خلي بالك الداتا بتاعه الملف كلها اتمسحت علشان استخدمنا وضع write
```

## JSON FILE
--------------------------

- first - import json
- reading json file :
-       json.load()

```python
import json

# بنفتح الملف بنفس الطريقة المعتادة (القاعدة الذهبية: utf-8)
with open('data.json', 'r', encoding='utf-8') as file:
    # load بتشفط الداتا من الملف وتحولها لـ Dictionary
    data = json.load(file)

# دلوقتي نقدر نتعامل معاها كأنها قاموس عادي جداً
print(data['name'])
```

writing in json file :
-       json.dump

```python
import json

# دي الداتا بتاعتنا جوه بايثون (Dictionary متداخل)
user_data = {
    "id": 101,
    "name": "Ahmed",
    "skills": ["Python", "SQL", "Airflow"], # لستة جوه القاموس!
    "is_active": True
}

with open('output.json', 'w', encoding='utf-8') as file:
    # dump بتاخد الداتا، وبعدين الملف اللي هتكتب فيه
    # indent=4 بتعمل مسافات وتنسيق عشان الملف يكون مقروء للعين البشرية
    json.dump(user_data, file, indent=4)
    
print("JSON file created successfully!")
```

- load   : File → Python
- loads  : String → Python

- dump   : Python → File
- dumps  : Python → String

Imagine you have a folder containing 50 CSV files and you want to read them all.This function returns a list of the names of all the files inside the folder.
-     os.listdir

```python
import os

folder_path = r"D:\Excel Projects\Data"

# هترجع لستة بأسماء كل الملفات
files = os.listdir(folder_path)

for file in files:
    # بنشيك إننا بناخد ملفات الـ csv بس
    if file.endswith('.csv'):
        print(f"Processing: {file}")
```

```python
from pathlib import Path

# بص الشياكة في لزق المسارات!
folder = Path("D:/Excel Projects/Data")
file_path = folder / "sales.csv"

print(file_path)
```


-     Does the file exist?

-     Inventory of the contents of the folder (The Power of glob)

```python
folder = Path("D:/Excel Projects/Data")

# هيجيب ملفات الـ CSV بس!
for csv_file in folder.glob('*.csv'):
    print(f"Processing: {csv_file.name}")
```

- .name   >> return    file name with path
- .stem   >> return    file name only 
- .suffix >> return    path only 
- .parent >> return    Get the folder containing this file
