# Tuples ()
# [1] Tuple Items Are Enclosed in Parentheses
# [2] You Can Remove The Parentheses If You Want
# [3] Tuple Are Ordered, To Use Index To Access Item
# [4] Tuple Are Immutable => You Cant Add or Delete
# [5] Tuple Items Is Not Unique
# [6] Tuple Can Have Different Data Types
# [7] Operators Used in Strings and Lists Available In Tuples

```python
# syntax and type of tuple

tupleOne = ('mohamed', 23)
tupleTwo = 'ahmed', 25

print(tupleOne)
print(tupleTwo)

print(type(tupleOne))
print(type(tupleTwo))
```

```python
# tuple indexing 

mytuple = ('mohamed', 23, 'ahmed', 25)
print(mytuple[0])    # prints 'mohamed'
print(mytuple[-1])   # prints 25
```

```python
# tuple assign values to variables

mytuple2 = ('osama', 30, 'ali', 35)
mytuple2 [0] = 'ahmed'  # This will raise an error because tuples are immutable
print(mytuple2)
```

```python
# tuple items 

mytuple3 = ('mohamed', 23, 'ahmed', 25, 'mohamed', True)   # items can be of any data type and can be repeated 'not unique'
print(mytuple3[-1])
```

```python
# tuple with one item is str 

mytuple4 = ('mohamed')   # tuple with one item is str
print(type(mytuple4))
mytuple5 = ('mohamed',)   # tuple with one item + comma is tuple  == mytuple5 = 'mohamed',   
print(type(mytuple5))
```

```python
# tuple concatenation

a = (1, 2, 3)
b = (4, 5, 6)
c = a + b
d = a + ('mohamed',) + b
print(c)   
print(d)
```

```python
# Tuple, List, String Repeat (*)

myString = "MO"
myList = [1, 2]
myTuple = ("A", "B")

print(myString * 2) 
print(myList * 2)    
print(myTuple * 2)   
```

# Tuple methods 
-----------------------------

* count () 

```python
myTuple = ("A", "B", "C", "D", "E", "A")
print(myTuple.count("A"))  
```

* index ()

```python
myTuple = ("A", "B", "C", "D", "E", "A")
print(myTuple.index("A"))   # returns the index of the first occurrence of the specified value
```

```python
# Tuple Destruct

a = ("A", "B", "C")

x, y, z = a

print(x)
print(y)
print(z)

print('*' * 50)

a = ("A", "B", 4, "C")
x, y, _, z = a  # ignore the third value by using an underscore (_)

print(x)
print(y)
print(z)
```

```python
# Check if Item Exists

colors = ("Red", "Green", "Blue")
print("Red" not in colors)
print("Red" in colors)
```

```python
# Tuple Unpacking

person = ("Mohamed", 22, "Egypt")

name, age, country = person

print(name)
print(age)
print(country)
```
