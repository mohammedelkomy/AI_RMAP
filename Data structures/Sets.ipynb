# Set
# [1] Set Items Are Enclosed in Curly Braces
# [2] Set Items Are Not Ordered And Not Indexed
# [3] Set Indexing and Slicing Cant Be Done
# [4] Set Has Only Immutable Data Types (Numbers, Strings, Tuples) List and Dict Are Not
# [5] 




```python
# Not Ordered And Not Indexed

mySetOne = {"Osama", "Ahmed", 100}
print(mySetOne)
print(mySetOne[0])  # This will raise a TypeError because sets are not indexed and slicing is not supported.
```

```python
# Has Only Immutable Data Types

mySetThree = {"Osama", 100, 100.5, True, [1, 2, 3]}

print(mySetThree) # This will raise a TypeError because sets can only contain immutable (hashable) data types, and lists are mutable.
```

```python
# Set Items Is Unique

myset = {"Osama", "Ahmed", 100, 100, 100.5, True, True}
print(myset)  # This will print the set with unique items only, duplicates will be removed.
```

# set methods
----------------------------

* clear 
- remove all items 

```python
myset = {"Osama", "Ahmed", 100, 100, 100.5, True, True}
myset.clear()
print(myset)  # This will print an empty set after clearing all items.
```

* union

```python
a = {1, 2, 3}
b = {3, 4, 5}
c = {6,7,8}
print(a.union(b, c))  # This will print the union of sets a and b, which is {
print (a | b | c)  # This will also print the union of sets a and b, which is {1, 2, 3, 4, 5}
```

* add 

```python
a = {1, 2, 3}
a.add(5)    # This will add the element 5 to the set a.
a.add(5,6)  # set.add() takes exactly one argument (2 given)
print(a)
```

* copy ()

```python
a = {1, 2, 3}
b = a.copy()  # This will create a shallow copy of set a and assign it to set b.
print(b)  # This will print the copied set b, which is {1, 2, 3}.
```

* remove ()

```python
a = {1, 2, 3}
b = a.remove(2)  # This will remove the element 2 from the set a.
print(a)        
```

* discard ()

```python
a = {1, 2, 3}
a.discard(6)  # This will attempt to remove the element 6 from the set a, but since 6 is not present, it will do nothing and not raise an error.
print(a)  # This will print the set a after attempting to remove the element 6, which is {1, 2, 3}.
```

* update 
* like union 

```python
a = {1, 2, 3, 5}
b = {3, 4, 5}
a.update(b)
print(a)
```

- difference

```python
a = {1, 2, 3, 5, 4}
b = {3, 4, 5}
print(a.difference(b))   # This will print the difference between sets a and b, which is {1, 2}. ==> elemnts that are in a but not in b.
print(a - b)             # same as above 
```

- difference_update ()

```python
c = {1, 2, 3, 4}
d = {1, 2, 3, "Osama", "Ahmed"}
print(c)
c.difference_update(d)  #  # This will  the difference between sets a and b with update the original set 'c'
print(c)
```

# intersection()


```python
c = {1, 2, 3, 4}
d = {1, 2, 3, "Osama", "Ahmed"}
print(c.intersection(d))   # c & d
print(c & d)
```

# intersection_update()

```python
c = {1, 2, 3, 4}
d = {1, 2, 3, "Osama", "Ahmed"}
print(c)
print(c.intersection_update(d))  # This will  the same values between sets c and d with update the original set 'c'
print(c)
```

* # symmetric_difference()

```python
i = {1, 2, 3, 4, 5, "X"}
j = {"Osama", "mohamed", 1, 2, 4}
print(i)
print(i.symmetric_difference(j))     # i ^ j  the output is the values that are not in i and j 
print(i)
```

* # symmetric_difference_update()

```python
i = {1, 2, 3, 4, 5, "X"}
j = {"Osama", "mohamed", 1, 2, 4}
print(i)
i.symmetric_difference_update(j)     # i ^ j  the output is the values that are not in i and j 
print(i)
```

```python
# issuperset()

a = {1, 2, 3, 4}
b = {1, 2, 3}

print(a.issuperset(b))
print(b.issuperset(a))
print(a.issuperset(a))

# True
```

```python
# issubset()

d = {1, 2, 3, 4}
e = {1, 2, 3}

print(d.issubset(e)) 
print(e.issubset(d)) 
print(e.issubset(e)) 
```

```python
# isdisjoint()  منفصلين عن بعض 

g = {1, 2, 3, 4}
h = {1, 2, 3}
a= {'mohamed', 1}
b= {'ahmed'}
print(g.isdisjoint(h))
print(g.isdisjoint(a))
print(g.isdisjoint(b))
print(type(b))
```
