# Nested Structures 
-----------------------------------

* النوع الاول وهو list in list 

```python
students = [
    ['Mohamed', 23],
    ['Ahmed', 26]
]

print(students[0])
print(students[0][1])
print(students[1][1])
```

* Dictionary in Dictionary

```python
students = {
    "Mohamed": {
        "age": 23,
        "grade": 90
    },
    "Ahmed": {
        "age": 22,
        "grade": 85
    }
}

print(students["Mohamed"])
print(students["Mohamed"]["age"])
print(students["Ahmed"]["grade"])
```

* list in Dictionary

```python

students = {
    "Mohamed": [90, 85, 95],
    "Ahmed": [80, 88, 75]
}

print(students["Mohamed"][-1])
print(students["Ahmed"][2])
```

dictionary in list  >> الشائع

```python

students = [
    {"name": "Mohamed", "age": 23},
    {"name": "Ahmed", "age": 22},
    {"name": "Ali", "age": 21}
]

print(students[1]["name"])
print(students[2]["age"])
```

```python

students = [
    {"name": "Mohamed", "grade": 90},
    {"name": "Ahmed", "grade": 85},
    {"name": "Ali", "grade": 78}
]

for student in students : 
    print(f" student name: {student['name']} \n with grade: {student['grade']} \n")
```
