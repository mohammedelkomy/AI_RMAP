- دي مكتبة مدمجة في بايثون  بتتعامل مع التاريخ والوقت، زي:
- معرفة تاريخ النهارده.
- معرفة الوقت الحالي.
- حساب الفرق بين تاريخين.
- إضافة أيام أو طرح أيام من تاريخ معين.

```python
from datetime import datetime as dt

print(dt.now())
print(dt.now().date())
print(dt.now().time())
```

```python
print(dt.now().strftime("%d-%m-%Y"))
print(dt.now().strftime("%I-%M %p"))       # strftime() change type from datetime to str
```

- %d > day
- %m > month
- %Y > year 
- %H > hours in (24) 
- %I > hours (12)
- %M > Minutes
- %S > seconds
- %p > AM/PM

```python
from datetime import date

today = date.today()
future_date = date(2026, 12, 31)

difference = future_date - today

print(difference.days)
```
