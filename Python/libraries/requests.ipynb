```python
# get()

import requests

response = requests.get("https://jsonplaceholder.typicode.com/users/1")   # بيبعت طلب GET لجلب البيانات.

print(response.status_code)    # بيوضح نتيجة الطلب؛ مثلًا 200 معناها إن الطلب نجح
print(response.text)           # بيرجع محتوى الاستجابة كنص.
```

```python
import requests

response = requests.get("https://jsonplaceholder.typicode.com/users/1")

user = response.json()      # بتحوّل محتوى الاستجابة إلى بيانات Python

print(user["name"])
print(user["email"])
```

```python
# التعامل مع اخطاء الاتصال 

import requests

try:
    response = requests.get(
        "https://jsonplaceholder.typicode.com/users/1",
        timeout=10      # مهلة الطلب
    )

    response.raise_for_status()             # بيرفع استثناء لو استجابة HTTP فيها خطأ
    user = response.json()                  # بيساعدك تتعامل مع أخطاء الطلبات والاتصال

    print(user["name"])

except requests.exceptions.RequestException as error:
    print(f"Request failed: {error}")
```

```python
# training 

import requests

try: 
    response = requests.get("https://jsonplaceholder.typicode.com/users/2")
    print(response.status_code)
    user = response.json()
    print(type(user))
    print(f"user_name is {user['name']} with email {user['email']}")
except requests.exceptions.RequestException as error : 
    print(error)
```

```python
import requests

try:
    response = requests.get(
       "https://httpbin.org/json",
        timeout=15
    )
    response.raise_for_status()

    data = response.json()
    print(response.status_code)
    print(type(data))
    print(f"slideshow is {data['slideshow']['title']} ")

except requests.exceptions.RequestException as error:
    print(error)
```
