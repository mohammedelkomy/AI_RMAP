مثلاً، بدل ما تفتح فولدر بنفسك وتعرف مكانه، تقدر تخلي بايثون يجيبلك المسار أو ينشئ فولدر جديد.

```python
# معرفة مكان التشغيل الحالي

import os 
print(os.getcwd())
```

```python
# عرض الملفات والمجلدات

files = os.listdir() 
print(files)

# files = os.listdir('you can write file path here') 
# print(files)
```

```python
# انشاء فولدر 

os.mkdir("MyProject")
```

```python
# التأكد من وجود ملف او فولدر

if os.path.exists("test.py"):
    print("Found")
else:
    print("Not found")


if os.path.exists("MyProject"):
    print("Found")
else:
    print("Not found")
```

```python
# حذف ملفات 

# os.remove("test.py")        # use .remove for files 
os.rmdir("MyProject")       # use .remove for empty folders
```

```python
# التعامل مع المسارات

folder = "data"
filename = "report.txt"

path = os.path.join(folder, filename)
print(path)
```
