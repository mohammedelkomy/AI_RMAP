
* اهم 4 دوال : 

    - **json.dumps** : 
        - converting Python data into text in JSON format

    - **json.loads** : 
        - converting the JSON text to the appropriate data type in Python
    

- **json.dumps** **===>** PythonObject to jsonString
- **json.loads** **===>** jsonString to pythonObject

 - **json.dump** : 
    - This code will create a file named employee.json and write the data inside it.

- **json.load** : 
    - It reads data from a JSON file and converts it into Python data.

* load, dump : To handle files
* loads, dumps : To deal with texts

```python
# json.dumps()          py > js

import json

employee = {
    "name": "Ahmed",
    "age": 25,
    "salary": 8000
}

data = json.dumps(employee)
data2 = json.dumps(employee, indent=4)

print(data)
print(type(data))
print(data2)
print(type(data2))
```

```python
# json.loads()       js > py

data = '{"name": "Ahmed", "age": 25, "salary": 8000}'

employee = json.loads(data)

print(employee)
print(type(employee))
print(employee["name"])
```

```python
# json.dump()          py in js


employee = {
    "name": "Ahmed",
    "age": 25,
    "salary": 8000
}

with open("employee.json", "w") as file:
    json.dump(employee, file, indent=4)
```

```python
# json.load ()             js > py


with open("employee.json", "r") as file:
    employee = json.load(file)

print(employee)
print(employee['name'])
```
