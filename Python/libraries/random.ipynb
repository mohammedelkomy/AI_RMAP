* بتساعدك تولّد أرقام عشوائية، وتختار عناصر عشوائية من قوائم، وتخلط ترتيب العناصر


```python
# random.randint(a,b)

import random 
print(random.randint(10, 20))  # 10 and 20 included
```

```python
# random.random() 

print(random.random())   # float number from 0 to 1
```

```python
# random.choise()  

names = ['Ahmed', "mohamed", 'sayed', '3bbas'] 
print(random.choice(names))   # choose a random item from a set
```

```python
# random.shuffle()

names = ['Ahmed', "mohamed", 'sayed', '3bbas'] 
random.shuffle(names)     # بتخلط ترتيب العناصر داخل القائمة نفسها
print(names)
```

```python
# training 

import random as r 

names = ['sayed', '3bbas', 'mohamed', 'Ahmed', 'Mahmoud']

print(r.choice(names))
print(r.randint(1, 100))

r.shuffle(names) 
print(names)
```
