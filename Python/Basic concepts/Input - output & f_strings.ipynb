```python

# 1. Output: print()
# بنستخدم print() علشان نعرض بيانات على الشاشة

print("Hello World")
print(10)
print("My name is Mohamed")

# Print multiple values
name = "Mohamed"
age = 22

print(name, age)


# 2. Input: input()
# بنستخدم input() علشان ناخد بيانات من المستخدم

name = input("Enter your name: ")
print(name)

# Input returns a string by default
age = input("Enter your age: ")
print(age)

# Convert input to int
age = int(input("Enter your age: "))

# Convert input to float
height = float(input("Enter your height: "))


# 3. f-Strings
# بنستخدم f-string علشان نحط المتغيرات داخل النص بسهولة

name = "Mohamed"
age = 22

print(f"My name is {name}")
print(f"My age is {age}")

# Use multiple variables
print(f"My name is {name} and I am {age} years old.")


# 4. f-Strings with calculations
x = 10
y = 5

print(f"Sum = {x + y}")
print(f"Multiply = {x * y}")


# 5. Complete example
name = input("Enter your name: ")
age = int(input("Enter your age: "))
height = float(input("Enter your height: "))

print(f"My name is {name}.")
print(f"I am {age} years old.")
print(f"My height is {height} meters.")
```
