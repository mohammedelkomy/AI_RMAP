# Lists [] 
# =============
# [1] List Items Are Enclosed in Square Brackets
# [2] List Are Ordered, To Use Index To Access Item ==> " starts by 0 " 
# [3] List Are Mutable => Add, Delete, Edit  ==> you can doing operations in lists
# [4] List Items Is Not Unique  ==>  [0,1,0,1,1]
# [5] List Can Have Different Data Types  ==> ['Mohamed', 0, True]
-----------------------------------------

```python
myList = [1, 2, 3, 5, 6, 7, 8, 9, 10]
print(f" #1 {myList} ")
print(f" #2 {myList[1]} ")
print(f" #3 {myList[-1]} ")
print(f" #4 {myList[1:4:2]} ")
print(f" #5 {type(myList)} ")
```

```python
# Slicing 
myList = [1,2,3,4,5]
print(myList[1:4])   # starts at index 1 and goes up to but does not include index 4 ==> we start from 0 

myList = [1,2,3,4,5,[6,7,8]]
print(myList[5])
print(myList[5][1])
```

# Lists Mehods 
---------------------------

*  append() 
* تُضيف هذه الطريقة عنصراً إلى نهاية القائمة. يمكن أن يكون العنصر الجديد من أي نوع بيانات، بما في ذلك قائمة أخرى.


```python
myList = [1,2,3,4,5]
myList.append(6)
print(myList)

print("*" * 50)

myList.append(['Mo',"OS"])
print(myList)
```

* extend() 
* تُستخدم هذه الطريقة لإضافة عناصر متعددة إلى نهاية القائمة. وهي تأخذ كائناً قابلاً للتكرار (مثل قائمة، أو صف، أو سلسلة نصية) كوسيط، وتضيف كل عنصر من عناصر ذلك الكائن إلى نهاية القائمة.

```python
a = [1, 2, 3, 4]
b = ["A", "B", "C"]

a.extend(b)
print(a)
```

* remove () 
* بتشيل عنصر معيم من الليست بتاعتي 

```python
x = [1, 2, 3, 4, 1,  'A', 'B', 'C']   # List with duplicate elements
x.remove(1)  # removes the first occurrence of 1 onlyyy
print(x)
```

* sort ()
* تستخدم لترتيب الليست

```python
y = [1, 2, 100, 120, -10, 17, 29]
y.sort()  # sorts the list in ascending order
print(y)

print("*" * 50)

y.sort(reverse=True)  # sorts the list in descending order
print(y)
```

* reverse () 
* بشقلب الترتيب بخلي اللي ورا قدام والعكس 


```python
z = [10, 1,9,80, 100, "Osama", 100]
z.reverse()  # reverses the list
print(z)
```

* clear ()
* مسح جميع العناصر من الليست

```python
z = [10, 1,9,80, 100, "Osama", 100]
z.clear()  # clears the list
print(z)
```

* copy ()
* باخد نسخة من الليست اللي عندي و مبتتاثرش لو غيرت اي حاجه في الليست الاصليه 

```python
z = [10, 1,9,80, 100, "Osama", 100]
p = z.copy()  # copies the list
print(p)

z= [1, 2, 3, 4, 5]
print(z)
print(p)
```

* count () 
* تستخدم لعد عنصر معين في الليست

```python
z = [10, 1,9,80, 100, "Osama", 100]
print(z.count(100))  # counts the number of occurrences of 100 in the list
```

* index ()
* بستخدمها علشان اعرف مكان حاجه معينه في الليست

```python
z = [10, 1,9,80, 100, "Osama", 100]
print(z.index(100))  # returns the index of the first occurrence of 100 in the list
```

* insert (index, Value)
* تستخدم لاضافه عنصر مع تحديد مكان الاضافه 

```python
z = [10, 1,9,80, 100, "Osama", 100]
z.insert(2, "Python")  # inserts "Python" at index 2
print(z)    
```

* pop () 
* بتجيب القيمه اللي في الليست بناءا علي قيمه الانديكس اللي انا عايزة

```python
z = [10, 1,9,80, 100, "Osama", 100]
print(z.pop(5))  # هنا هترجعلي القيمه اللي في الاي اندكس 5 و هتشيلها من الليست
print(z)  # الليست بعد الحذف ل الاي اندكس 5

print("*" * 50)

# اشيل اخر رقم من الليست وطباعه الليست بعد الحذف
z = [10, 1,9,80, 100, "Osama", 100]
z.pop()
print(z)
```
