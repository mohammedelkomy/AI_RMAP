# Dictionaries
# [1] Dict Items Are Enclosed in Curly Braces
# [2] Dict Items Are Contains Key : Value
# [3] Dict Key Need To Be Immutable => (Number, String, Tuple) List Not Allowed
# [4] Dict Value Can Have Any Data Types
# [5] Dict Key Need To Be Unique
# [6] Dict Is Not Ordered You Access Its Element With Key
-------------------------------------------

```python
user = {
    "name": "Mohamed",
    "age": 36,
    "country": "Egypt",
    "skills": ["Html", "Css", "JS"],
    "rating": 10.5
}

print(user)
print(user.keys())
print(user.values())
print(user.items())
```

```python
# access elemnt with key 

print(user['country'])
print(user.get('country'))
```

```python
print(user.keys())
print(user.values())
```

```python
# Two-Dimensional Dictionary  >> nested dict 

languages = {
  "One": {
    "name": "Html",
    "progress": "80%"
  },
  "Two": {
    "name": "Css",
    "progress": "90%"
  },
  "Three": {
    "name": "Js",
    "progress": "90%"
  }
}

print(languages)
print(languages['One'])
print(languages['Three']['name'])

```

```python
# Dictionary Length

print(len(languages))
print(len(languages["Two"]))
```

```python

# Create Dictionary From Variables

frameworkOne = {
  "name": "Vuejs",
  "progress": "80%"
}

frameworkTwo = {
  "name": "ReactJs",
  "progress": "80%"
}

frameworkThree = {
  "name": "Angular",
  "progress": "80%"
}

allFramework = {
  "one": frameworkOne,
  "two": frameworkTwo,
  "three": frameworkThree
}

print(allFramework)
```

* Dictionary Methods 
-----------------------------


```python
# clear()  ====>  remove all items 

user = {
  "name": "Osama"
}
print(user)
user.clear()
print(user)



```

```python
# update()

member = {
  "name": "Osama"
}
print(member)
member["age"] = 36
print(member)
member.update({"country": "Egypt"})
print(member)


```

```python
# copy()

main = {
  "name": "Osama"
}

b = main.copy()
print(b)
main.update({"skills": "Fighting"})
print(main)
print(b)

```

```python
# keys() + values()

print(main.keys())
print(main.values())
```

```python
# setdefault()

user = {
  "name": "Osama"
}
print(user)
print(user.setdefault("age", 36))
print(user)

print("=" * 40)

```

```python
# popitem()

member = {
  "name": "Osama",
  "skill": "PS4"
}
print(member)
member.update({"age": 36})
print(member.popitem())
print(member)
print("=" * 40)

```

```python
# items()

view = {
  "name": "Osama",
  "skill": "XBox"
}

allItems = view.items()
print(view)
view["age"] = 36

print(allItems)

print("=" * 40)

```

```python
# fromkeys()

a = ('MyKeyOne', 'MyKeyTwo', 'MyKeyThree')
b = "X"

print(dict.fromkeys(a, b))
```

```python
student = {
    "name": "Mohamed",
    "age": 22,
    "grade": 85.5
}

for key , value in student.items():
    print(f" {key} : {value} ")
```

افتكر:
- الـ Key لازم يكون فريدًا داخل الـ Dictionary؛ لو كررت المفتاح، القيمة الأخيرة هي اللي بتفضل.
- student["name"] بتجيب القيمة المرتبطة بالمفتاح.
- get() بتساعدك تتجنب KeyError لو المفتاح مش موجود.
- تقدر تضيف وتعدّل وتحذف بيانات.
- items() بتتيح لك التعامل مع المفتاح والقيمة مع بعض.
