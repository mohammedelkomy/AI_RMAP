# *args 
# تخيّل إنك بتبني فانكشن بتجمع أرقام، ومش عارف المستخدم هيدخل كام رقم

```python
def add_numbers(*args):
    print(args)

add_numbers(10, 20, 30)
```

```python
def add_numbers(*args):
    total = 0

    for number in args:
        total += number

    return total

print(add_numbers(10, 20, 30))
print(add_numbers(5, 15))
```

```python
def show_datails (name, *skills) : 
    print(f"Hello {name} your skills is :")
    for skill in skills : 
        print (f"- {skill}")

show_datails("Mohamed",'CSS',"Py",'excel')
```

# **kwargs
# دلوقتي تخيّل إنك بتعمل فانكشن تستقبل بيانات موظف، لكن مش عارف عدد البيانات أو أسماء الحقول اللي هتتبعت

```python
def show_employee(**kwargs):
    print(kwargs)

show_employee(name="Ahmed", salary=5000, department="IT")
```

```python
def show_employee(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_employee(name="Ahmed", salary=5000, department="IT")
```

```python
def example(*args, **kwargs):
    print(args)
    print(kwargs)

example(10, 20, name="Mohamed", age=23)
```

* Flexible Employee Report


```python
def employee_report (*args, **kwargs): 
    total_salaries = 0
    count = 0

    for salary in args : 
        total_salaries += salary
        count += 1 

    print(f"Total salaries: {total_salaries}")
    print(f"Average salaries:", total_salaries/count)

    for key, value in kwargs.items():
        print(key,value)

    if args == 0 or args == "": 
        pass

employee_report(
    4000, 6000, 8000,
    name="Mohamed",
    department="IT",
    city="Cairo"
)
```
