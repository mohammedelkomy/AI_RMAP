# Lambda
# -------
# [1] It Has No Name
# [2] You Can Call It Inline Without Defining It
# [3] You Can Use It In Return Data From Another Function
# [4] Lambda Used For Simple Functions and Def Handle The Large Tasks
# [5] Lambda is One Single Expression not Block Of Code
# [6] Lambda Type is Function
---------------------------------

```python
def hello(name,age) : 
    return f"Hello {name} your age is {age}" 

print(hello("Mohamed",23))
```

```python
# in lambda func

hello = lambda name, age : f'Hello {name} your age is {age}'
print(hello('Ahmed',26))
```

--------------------------------------------------

# Map 
# [0]Built In Functions
# [1] Map Take A Function + Iterator
# [2] Map Called Map Because It Map The Function On Every Element
# [3] The Function Can Be Pre-Defined Function or Lambda Function
----------------------------

* بتطبيق عملية علي كل عنصر 
* map (function, iterable)


```python
salaries = [1200, 4500, 5000]

new_salaries = map(lambda salary: salary + 500, salaries) 

print(new_salaries)
print(list(new_salaries))
```

----------------------

# Filter () >> اختيار عناصر بشروط 
# filter (function, iterable)
--------------------------


```python
# find salary > 4000 

salaries = [3000, 4000, 5000, 2500, 7000]

result = list(filter(lambda salary: salary > 4000, salaries))

print(result)
```

---------------------------

# Sorted() >> ترتيب العناصر
# sorted (iterable)
---------------------------

```python
salaries = [3000, 4000, 5000, 2500, 7000]

print(sorted(salaries))                  # ASC

print(sorted(salaries, reverse=True))    # DESC
```

---------------------------------------
# key >> تحديد الترتيب هيتم علي اساس ايه 
------------------------

```python
employees = [
    {"name": "Ahmed", "salary": 5000},
    {"name": "Mona", "salary": 8000},
    {"name": "Ali", "salary": 3000}
]

result = sorted(employees, key= lambda employee: employee['salary'])
print(result)
```

```python
employees = [
    {"name": "Ahmed", "salary": 3500},
    {"name": "Mona", "salary": 7000},
    {"name": "Ali", "salary": 4500},
    {"name": "Sara", "salary": 9000}
]
```

```python
# add 500 for each salary 

new_salary = list(map(lambda employee: employee['salary'] + 500, employees))
print(new_salary)
```

```python
# find employees with salary > 4000 

result = list(filter(lambda employee: employee['salary'] > 4000, employees))
print(result)
```

```python
# sort employee based in salary desc 
sorting = sorted(employees, key=lambda employee: employee['salary'], reverse=True)
print(sorted(employees, key=lambda employee: employee['salary'], reverse=True))
print(type(sorting))
```
