```python
# Employee Salary Calculator

def calculate_salary (salary, bonus=0) : 
    if salary <= 5000 : 
        total_salary = salary + 500
    else: 
        total_salary = salary + bonus

    return total_salary

print(calculate_salary(6000,1500))
print(calculate_salary(4500))
```

```python

employees = [
    {"name": "Ahmed", "salary": 4500, "bonus": 500},
    {"name": "Mona", "salary": 7000, "bonus": 1000},
    {"name": "Ali", "salary": 3500, "bonus": 200},
    {"name": "Sara", "salary": 9000, "bonus": 1500}
]
print(employees[0]['salary'])
```

```python
for employee in employees:
    print(f"{employee['name']} with {employee['salary']}")
```

```python
def calculate_salary (salary, bonus=0) : 
    if salary < 4000 : 
        total_salary = salary + bonus + 500

    elif  8000 > salary >= 4000 : 
        total_salary = salary + bonus 

    else: 
        total_salary = salary + bonus + (.2 * salary)

    return total_salary

print(calculate_salary(6000,250))
```

```python
def process_employee (employees): 

    total_salaries = 0
    employee_count_6000 = 0
    max_salary = 0
    max_salary_name = ""


    for employee in employees : 
        total_salary =  calculate_salary(employee['salary'],employee['bonus'])
        print(f'Name is: {employee['name']} with total_salary: {total_salary}' ) 

        total_salaries += total_salary

        if total_salary > 6000 : 
            employee_count_6000 += 1 

        if total_salary > max_salary : 
            max_salary = total_salary
            max_salary_name = employee['name']
        
    print(f'total_salaries is: {total_salaries}')
    print(employee_count_6000)
    print(f"Higest Salary Person is {max_salary_name} with salary : {max_salary}" )

process_employee(employees)
```

* Function Recursion

```python
def cleanWord(word):
    if len(word) == 1:
        return word

    if word[0] == word[1]:
        return cleanWord(word[1:])

    return word[0] + cleanWord(word[1:])

print(cleanWord('KKOOMMYYY'))
```
