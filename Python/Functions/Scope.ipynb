# Scope ==> Local & Global
* هو النطاق اللي المتغير بيكون معروف ومتاح للاستخدام جواه
* يعني لما تعرّف متغير، مش شرط تقدر تستخدمه من أي مكان في البرنامج. ده بيعتمد على المكان اللي اتعرّف فيه.
------------------------------

* LOCAL SCOPE 

```python
def my_function():
    message = "Hello Mohamed"  # local
    print(message)

my_function()
```

```python
def my_function():
    message = "Hello Mohamed"

my_function()
print(message)
```

* GLOBAL SCOPE

```python
name = "Mohamed"

def greet():
    print(name)

greet()
print(name)
```

```python
count = 10

def update_count():
    count = 5
    print(count)   # return loacl 

update_count()    # local
print(count)      # global
```

* لو عايز تغيّر المتغير العام من داخل Function، تقدر تستخدم global.

```python
count = 10

def update_count():
    global count
    count = 5

update_count()
print(count)
```
