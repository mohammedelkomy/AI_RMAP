# Exception Handling
--------------------------
* try: حط جواه الكود اللي ممكن يحصل فيه خطأ.
* except: لو حصل خطأ من النوع اللي بتتعامل معاه، نفّذ الكود الموجود جواه بدل ما البرنامج يقف بسبب الخطأ ده
* finally معناها إن فيه كود هيتنفذ في النهاية، سواء حصل خطأ أو مفيش خطأ،

```python
x= '10'
if type(x) != int: 
    raise ValueError ("only integer allowed")
else:
    print(x/2)
```

```python
try : 
    salary = int(input("Please Enter your salary: "))
    print(f'Your salary + bonus is {salary+500}')
except : 
    print('please enter valid number')
finally: 
    print("program finished")
```

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
    print(result)

except ValueError:
    print("You must enter a number")

except ZeroDivisionError:
    print("You cannot divide by zero")

finally:
    print("Done")

# 'mohamed' >> "You must enter a number"
# 0 >> You cannot divide by zero
# 20 >> 5
```

```python
try : 
    salary = int(input("enter salary: "))
    if salary > 0 : 
        print(salary)
    else: 
        print("salary cannot be zero or less")
except ValueError: 
    print("Invalid salary")
finally:
    print("salary check completed")

```

```python
try:
    n1= float(input("enter fisrt number: "))
    n2= float(input("enter second number: "))

    print(f"{n1}/{n2} = {n1/n2}") 

except ValueError : 
    print("enter valid numbers")

except ZeroDivisionError : 
    print("second number cannot be zero")

finally:
    print("calculation finished")
```

```python
employees = [
    {"name": "Mohamed", "salary": 8000},
    {"name": "Ahmed", "salary": 6000},
    {"name": "Ali", "salary": 7000}
]
```

```python
try: 
    e_name = str(input("Enter Emp Name: "))
    found = False

    for employee in employees : 
        if e_name.lower() == employee['name'].lower() : 
            print(f"name is: {employee['name']} with salary: {employee['salary']}")
            found = True
            break

    if not found : 
        print(f"{e_name} not found")

finally:
    print('search completed')
```
