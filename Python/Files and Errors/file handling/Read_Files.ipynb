#  File Handling => Read File 
 --------------------------------

```python
myfile = open(r"D:\test.txt",'r')

print(myfile.read())
```

```python
myfile.seek(0)   # return pointer to zero 

for line in myfile :
    print(line.upper())
```

```python
# best way to read file : 

with open(r"D:\test.txt",'r') as myfile :
    print(myfile.read())
    
```
