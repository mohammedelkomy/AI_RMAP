#  File Handling => Write & Append File 
 --------------------------------

* Write >> Remove Data and insert new data 

```python
file = open(r"D:\testt.txt",'w') 
file.write('Heloo')
file.close()
```

```python
with open(r"D:\testt.txt",'r') as file : 
    for line in file : 
        print(line)
```

```python
with open(r"D:\testt.txt",'w') as file: 
    file.write("Hello Mohamed \n")
    file.write('how are you :)')
```

```python
with open(r"D:\testt.txt",'r') as file : 
    for line in file : 
        print(line)
```

```python
# write lines

mylist = ['Mohamed\n', 'Ahmed\n', 'Mahmoud\n'] 
with open(r"D:\testt.txt",'w') as file : 
    file.writelines(mylist)


with open(r"D:\testt.txt",'r') as file : 
    for line in file : 
        print(line)
```

-----------------------

* append >> add new data to old data without removing 

```python
with open(r"D:\testt.txt",'a') as file: 
    file.write("Sameh")


with open(r"D:\testt.txt",'r') as file : 
    for line in file : 
        print(line)
```
