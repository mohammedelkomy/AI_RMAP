----------------------------
# write with JSON file 
----------------------------

```python
import json 

employee = {
    "name" : 'Mohamed',
    "dept" : "Ai",
    "salary" : 8000
}

with open("employee.json", "w") as file : 
    json.dump(employee, file, indent=4)
```

* json.dump()   >> convert from py to json
* indent=4      >> make file readable 

----------------------------
# Read with JSON file 
----------------------------

```python
import json 

with open("employee.json", "r") as file : 
    employee = json.load(file) 

print(employee)
print(employee['name'])
print(employee['salary'])
```
