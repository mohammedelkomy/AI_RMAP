----------------------------
# write with csv file 
----------------------------

```python
import csv

with open('employees.csv', 'w', newline="") as file : 
    writer = csv.writer(file) 

    writer.writerow(["name", "departement", "salary"])
    writer.writerow(["ahmed", "it", "10000"])
    writer.writerow(["mohamed", "ai", "25000"])
```

----------------------------
# Read with csv file 
----------------------------

```python
import csv 

with open("employees.csv", "r", newline="") as file : 
    reader = csv.reader(file)

    for row in reader : 
        print(row)
```
