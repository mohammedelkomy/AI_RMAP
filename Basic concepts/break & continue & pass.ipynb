# 1. break
# بيوقف ال لوب بالكامل عند تحقيق شرط معين 
-----------------------------------------------------------

```python
numbers = [1, 2, 3, 4, 5]

for n in numbers : 
    
    if n == 3 : 
        break
    
    print(n)
```

# 2. continue
# بيتخطي التكرار الحالي وبكمل للي بعده 
---------------------------------------------------------------

```python
numbers = [1, 2, 3, 4, 5]

for n in numbers : 
    
    if n == 3 :       # 3 not printed because of continue statement 
        continue
    print(n)
```

# 3. pass
# معناها "عدّي من هنا من غير ما تعمل أي حاجة". بنستخدمها لما نكون محتاجين نكتب مكان كود لسه هننفذه بعدين، أو نسيب جزء فاضي مؤقتًا. 
------------------------------------------

```python
age = int(input("Enter your age : "))

if age >= 18 : 
    print("You are an adult you can Enter the club")
else : 
    pass
print("program ended")
```
