# Name Convention and Rules

# [1] Can Start With (a-z A-Z) Or Underscore
# [2] You Cannot With Num Or Special Characters
# [3] Can Include (0-9) Or Underscore
# [4] Cannot Include Special Characters
# [5] Name is Not Like name [ Case Sensitive ]

---------------------------------------------------

```python

# Variables

# Variable: مكان بنخزن فيه قيمة ونستخدمها بعد كده

name = "Mohamed"
age = 22
height = 1.70
is_student = False

# Data Types

# 1. String (str): نص
name = "Mohamed"

# 2. Integer (int): عدد صحيح
age = 22

# 3. Float: عدد عشري
height = 1.70

# 4. Boolean (bool): True or False
is_student = False

# Check Data Type
print(type(name))
print(type(age))
print(type(height))
print(type(is_student))

# Print Variables
print(name)
print(age)

# Combine Text with Variables
print("My name is", name)
print("My age is", age)

# Update Variable
age = 23
print(age)

# Get Input from User
user_name = input("Enter your name: ")
print(user_name)

# Input is String by default
user_age = int(input("Enter your age: "))

# Convert Data Types
x = int("10")
y = float("5.5")
z = str(100)

print(x)
print(y)
print(z)
```
