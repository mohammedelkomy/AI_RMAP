# For Loop 
# بنستخدمها لما نكون عايزين نكرر على عناصر أو عدد مرات محدد


```python
numbers = [1, 2, 3, 4, 5]  # data type is list
for num in numbers:
    print(num)
else : 
    print("No more numbers in the list.")  # التنفيذ على أي حال بعد انتهاء الحلقة
```

```python
for i in range(5): # التكرار عبر سلسلة من الأرقام من 0 إلى 4
    print(i)
else : 
    print('list is finished')  # التنفيذ على أي حال بعد انتهاء الحلقة
```

```python
for i in range (0,10):  # التكرار عبر سلسلة من الأرقام من 0 إلى 9
    print(i) 
```

```python
for i in range (0,10,2): # التكرار عبر سلسلة من الأرقام من 0 إلى 9 مع زيادة بمقدار 2
    print(i)
```

```python
name = "Mohamed"
for char in name:
    print(char)
```

# Practise 

```python
number = float(input("Enter a number: "))
total = 0
for a in range(1, int(number) + 1): 
    print(a) 
    total += a
print(f"Total = {total}")
```
