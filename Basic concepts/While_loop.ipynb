# While Loop 
بتتكرر طول ما الشرط بيتحقق.
---------------------

```python
a = 0 
while a <= 10:        # While condition is True Code will run 
    print(a)
    a += 2
else :               # excuted when the while condition is Fasle  ==> True become false
    print("a is no longer less than 10")
```

```python
names = ['Mohamed', 'Ali', 'Ahmed', 'Hassan']
a = 0 
while a < len(names) : 
    print(f"{a+1}. {names[a]}")
    a += 1
else : 
    print(">> All names printed >> ")
```

```python
password = ""
while password != "1234" : 
    password = input("Enter your password : ") 
else : 
    print("access granted")
```
