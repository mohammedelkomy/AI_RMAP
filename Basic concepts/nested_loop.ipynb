```python
peoples = {
    "Osama" : {
        "excel" : "90%",
        "sql" : "80%" 
     },
     "Ahmed" : {
        "excel" : "70%",
        "sql" : "60%" 
     }
}

# for name, skills in peoples.items():
#     print(f"Name: {name}")
#     for skill, value in skills.items():
#         print(f"{skill} : {value}")


print(f"skills and progress for : ")
for name in peoples:
    print(f">> {name} >> : ")
    for skill in peoples[name] : 
        print(f"{skill} ==> {peoples[name][skill]}")
```
