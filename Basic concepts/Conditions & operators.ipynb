# operators

```python

# 1. Arithmetic Operators
# العمليات الحسابية

x = 10
y = 3

print(x + y)   # Addition: الجمع
print(x - y)   # Subtraction: الطرح
print(x * y)   # Multiplication: الضرب
print(x / y)   # Division: القسمة العادية
print(x // y)  # Floor Division: القسمة بدون الجزء العشري
print(x % y)   # Modulus: باقي القسمة
print(x ** y)  # Exponentiation: الأس


# 2. Comparison Operators
# المقارنة: النتيجة بتكون True أو False

a = 10
b = 5

print(a == b)  # Equal to
print(a != b)  # Not equal to
print(a > b)   # Greater than
print(a < b)   # Less than
print(a >= b)  # Greater than or equal to
print(a <= b)  # Less than or equal to


# 3. Assignment Operators
# إسناد وتحديث قيمة المتغير

number = 10

number += 5  # number = number + 5
print(number)

number -= 3  # number = number - 3
print(number)

number *= 2  # number = number * 2
print(number)

number /= 4  # number = number / 4
print(number)


# 4. Logical Operators
# العمليات المنطقية

age = 22
has_id = True

print(age >= 18 and has_id == True)  # True
print(age < 18 or has_id == True)    # True
print(not has_id)                    # False


# 5. Membership Operators
# التحقق من وجود عنصر داخل مجموعة أو نص

name = "Mohamed"

print("M" in name)      # True
print("z" in name)      # False
print("z" not in name)  # True


# 6. Identity Operators
# التحقق هل متغيران يشيران لنفس الكائن

x = [1, 2, 3]
y = x
z = [1, 2, 3]

print(x is y)  # True
print(x is z)  # False
print(x == z)  # True
```
